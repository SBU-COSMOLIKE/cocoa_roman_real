# Plot the FLAMINGO baryon response emulator

Developer notebook that draws the response $P(k)/P_{\rm DMO}(k)$ predicted by
the FLAMINGO baryon response emulator (`FlamingoBaryonResponseEmulator`) at
11 redshifts from 0 to 2, for 121 wavenumbers between $10^{-1.5}$ and
$10^{1.5}\,h/\mathrm{Mpc}$ (DMO: dark-matter only). Three numbers set the
model: `fgas_sigma` shifts the cluster gas fractions and `Mstar_sigma` the
stellar mass function, both in units of the observational uncertainty the
FLAMINGO simulations were calibrated to, and `jet_fraction` interpolates
between thermal AGN feedback (0) and kinetic jets (1).

Inputs: no files; it needs the `FlamingoBaryonResponseEmulator` package. The
text drawn in the upper-left corner of the figure is fixed and does not
follow the three model values set in the cells.

```python
import FlamingoBaryonResponseEmulator as fre
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.pyplot import cm
```

```python
# Load FLAMINGO emulator
emulator = fre.FlamingoBaryonResponseEmulator()
```

```python
# Mode bins in [h / Mpc]
k_min = 10**-1.5
k_max = 10**1.5
num_bins_k = 121
bins_k = 10 ** (np.linspace(np.log10(k_min), np.log10(k_max), num_bins_k))
k_min_plot = 0.02
k_max_plot = 50
```

```python
# Bins of redshifts
redshifts = np.linspace(0.0, 2.0, 11)
```

```python
# Specify baryon model
fgas_sigma = 0.0
Mstar_sigma = 0.0
jet_fraction = 1.0
```

```python
# Prepare figure
ax = plt.subplot()
ax.set_xscale("log")
ax.hlines(1, 1e-4, 1e4, ls="-", color="k", lw=0.7)

colors_z = cm.plasma(np.linspace(0.0, 0.9, len(redshifts)))

# Plot emulator data
for i in range(np.size(redshifts)):

    # Call emulator
    response = emulator.predict(
        bins_k, redshifts[i], fgas_sigma, Mstar_sigma, jet_fraction
    )

    # Plot response
    ax.plot(
        bins_k,
        response,
        color=colors_z[i],
        label="$z=%.1f$" % redshifts[i],
        lw=2,
        ms=1.5,
    )

# Plot range
ax.set_xlim(k_min_plot, k_max_plot)
ax.set_ylim(0.72, 1.14)
ax.set_ylabel("$P(k) / P_{\\rm DMO}(k)~[-]$", labelpad=2)
ax.set_xlabel("${\\rm Mode}~k~[h\\cdot{\\rm Mpc}^{-1}]$", labelpad=0)

# Legend and model
legend = ax.legend(
    loc="lower left",
    fancybox=True,
    framealpha=1,
    handlelength=1,
    ncol=3,
    columnspacing=0.5,
    handletextpad=0.2,
)
legend.get_frame().set_edgecolor("white")
ax.text(
    k_min * 1.2,
    1.13,
    "${\\rm fgas}+0\\sigma$\n${\\rm M*}+0\\sigma$\n${\\rm JET}~0\\%$",
    va="top",
    ha="left",
)
```
