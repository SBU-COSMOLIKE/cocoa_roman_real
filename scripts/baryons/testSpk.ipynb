# Check the SP(k) and BCemu suppression curves

Developer notebook that evaluates two baryonic-feedback models outside the
`bfmt` theory block and compares their output with the wavenumber grid of
the CosmoLike likelihood.

1. **SP(k)** (the `pyspk` package): the suppression
   $S(k)=P_{\rm hydro}/P_{\rm DMO}$ at $z=0.3$ for the Akino et al. (2022)
   baryon-fraction relation ($\alpha=4.189$, $\beta=1.273$,
   $\gamma=0.298$; `SO=500`, halo masses within spheres of 500 times the
   critical density) in a flat $\Lambda$CDM cosmology with
   $H_0=67.32$ km/s/Mpc and $\Omega_m=0.3$.
2. **Interpolation**: $\log_{10}S$ is interpolated linearly in
   $\log_{10}k$ and evaluated on the likelihood's grid `log10k_interp_2D`
   (1,500 nodes from −4.99 to 2.0 at boost 1), with $S=1$ below
   $k=10^{-3}$.
3. **BCemu**: the seven-parameter baryonification boost at $z=1$ for
   $\Omega_b=0.05$ and $\Omega_m=0.27$.

Inputs: no files; it needs the `pyspk`, `astropy` and `BCemu` packages.
Units: `pyspk` and BCemu use $k$ in $h/\mathrm{Mpc}$, while the likelihood's
`log10k_interp_2D` is $\log_{10}$ of $k$ in 1/Mpc.

```python
import numpy as np
import scipy
from scipy.interpolate import interp1d
import pyspk as spk
```

```python
from astropy.cosmology import FlatLambdaCDM
import matplotlib.pyplot as plt
```

```python
cosmo = FlatLambdaCDM(H0 = 67.32, Om0 = 0.3)
alpha = 4.189
beta = 1.273
gamma = 0.298
```

```python
k_spk, sup = spk.sup_model(SO=500, z = 0.3, alpha = alpha, beta = beta, gamma = gamma, cosmo = cosmo, k_min = 0.001, verbose = True)
```

```python
print(k_spk)
```

```python
print(sup)
```

```python
plt.plot(k_spk, sup)
plt.xscale('log')
```

```python
accboost = 1
# The likelihood's log10 k grid (k in 1/Mpc); accboost is its accuracy boost
log10k_interp_2D = np.linspace(-4.99,2.0,int(1250+250*accboost))
```

```python
interp_spk = interp1d(np.log10(k_spk), 
                                  np.log10(sup), kind = 'linear', fill_value = 'extrapolate', assume_sorted = True)
```

```python
# lnbt_spk holds log10 S, because the interpolant works in log10
lnbt_spk = interp_spk(log10k_interp_2D)
```

```python
np.max(np.power(10, log10k_interp_2D))
```

```python
print(lnbt_spk)
```

```python
min(k_spk)
```

```python
# S = 1 (log10 S = 0) below k = 1e-3
lnbt_spk[np.power(10,log10k_interp_2D)<1e-3] = 0
```

```python
plt.plot(k_spk, sup)
plt.plot(np.power(10, log10k_interp_2D), np.power(10,lnbt_spk))
plt.xscale('log')
plt.xlim(1e-3, 10)
```

```python
import BCemu
```

```python
bfcemu = BCemu.BCM_7param(Ob=0.05, Om=0.27)
bcmdict = {'log10Mc': 13.32,
           'mu'     : 0.93,
           'thej'   : 4.235,  
           'gamma'  : 2.25,
           'delta'  : 6.40,
           'eta'    : 0.15,
           'deta'   : 0.14,
           }

z = 1
#k_eval = 10**np.linspace(-1,1.08,50)
k_eval = 10** log10k_interp_2D
p_eval = bfcemu.get_boost(z, bcmdict, k_eval)

plt.semilogx(k_eval, p_eval, c='C0', lw=3)
plt.axis([1e-3,100,0.73,1.04])
plt.yticks(fontsize=14)
plt.xticks(fontsize=14)
plt.xlabel(r'$k$ (h/Mpc)', fontsize=14)
plt.ylabel(r'$\frac{P_{\rm DM+baryon}}{P_{\rm DM}}$', fontsize=21)
plt.tight_layout()
plt.show()
```
